```python
#Pembuatan boxplot menggunakan seaborn

import matplotlib.pyplot as plt
import seaborn as sns
penguins = sns.load_dataset("penguins")
penguins.info()
```

```python
sns.boxplot(x=penguins['bill_length_mm'])
```

```python
sns.boxplot(x=penguins['species'],y=penguins['bill_length_mm'])
```

```python
sns.boxplot(data=penguins, x="species", y='bill_length_mm',
            color="red")
```

```python
sns.boxplot(data=penguins, x="species", y='bill_length_mm', palette="muted")
```

```python
custom_pal = {"Adelie": "#9A0EEA", "Chinstrap": "beige", "Gentoo":"c"}
sns.boxplot(data=penguins,
 x="species", y='bill_length_mm',
 palette=custom_pal,
 linewidth=1,
 width=0.4,
 notch=True,
 fliersize=10,
 flierprops={"marker": "o"})
```

```python
#Boxplot dengan Subkategori
# Mengatur ukuran frame
plt.figure(figsize=(8, 6))
sns.boxplot(data=penguins,
 x="species", y='bill_length_mm', hue="sex",
 palette="bright",
 linewidth=0.8,
 width=0.5,
 fliersize=3,
 flierprops={"marker": "o"})
```
