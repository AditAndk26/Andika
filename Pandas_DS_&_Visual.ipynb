```python
# Import utama
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- SERIES ---
print("Series penjualan mobil (unit):")
ds = pd.Series([120, 150, 90, 200], index=["Toyota","Honda","Mitsubishi","Suzuki"])
print(ds)

# Operasi Series
print("\nToyota + Honda:", ds["Toyota"] + ds["Honda"])
print("Rata-rata penjualan:", ds.mean())
print("Mobil terlaris:", ds.idxmax(), "dengan", ds.max(), "unit")

# --- DATA FRAME PENJUALAN MOBIL ---
data = {
    "bulan": [1,2,3,4,5,6],
    "Toyota":    [120, 135, 150, 160, 170, 180],
    "Honda":     [100, 110, 120, 140, 130, 135],
    "Mitsubishi":[80, 95,  100, 110, 115, 120],
    "Suzuki":    [90, 85,  95,  100, 105, 110]
}
df = pd.DataFrame(data)
print("\nDataFrame Penjualan Mobil:")
print(df)

# Statistik dasar
print("\nStatistik Toyota:")
print("Total:", df["Toyota"].sum(), "unit")
print("Rata-rata:", df["Toyota"].mean(), "unit/bulan")
print("Tertinggi:", df["Toyota"].max(), "unit")

# --- VISUALISASI ---
plt.figure(figsize=(8,5))
sns.lineplot(data=df.set_index("bulan"))
plt.title("Tren Penjualan Mobil per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Unit Terjual")
plt.grid(True)
plt.show()

# --- LOAD CSV (misal hasil_penjualan_mobil.csv) ---
# Simulasikan CSV kecil (jika ada file asli tinggal ganti path)
csv_data = {
    "bulan":[7,8,9,10],
    "Toyota":[190,200,210,220],
    "Honda":[140,145,150,155],
    "Mitsubishi":[125,130,135,140],
    "Suzuki":[115,120,125,130]
}
df_csv = pd.DataFrame(csv_data)
df_csv.to_csv("hasil_penjualan_mobil.csv", index=False)

# Baca kembali file CSV
df2 = pd.read_csv("hasil_penjualan_mobil.csv")
print("\nData dari CSV:")
print(df2)

# Gabungkan dengan data awal
df_all = pd.concat([df, df2], ignore_index=True)
print("\nGabungan DataFrame Penjualan Mobil (1-10 bulan):")
print(df_all)

# --- EKSTRAK & MANIPULASI ---
# Tambahkan kolom total penjualan per bulan
df_all["Total"] = df_all[["Toyota","Honda","Mitsubishi","Suzuki"]].sum(axis=1)
print("\nData dengan kolom Total:")
print(df_all.head())

# Urutkan bulan dengan penjualan total tertinggi
print("\nBulan dengan penjualan terbanyak:")
print(df_all.sort_values(by="Total", ascending=False).head(3))
```
