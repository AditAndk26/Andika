```python
# --- List contoh campuran ---
my_list = [15, "Fauzan", 9.5, True]
print("List awal:", my_list)

# Hapus elemen
my_list.remove(15)
print("Setelah remove:", my_list)

# Update elemen
my_list[1] = "Machine Learning"
print("Setelah update:", my_list)

# Tambah elemen baru
my_list.append("Bisdig")
print("Setelah append:", my_list)


# --- Array dengan modul array ---
import array as arr
my_array = arr.array('i', [1, 2, 3, 4, 5])
my_array.append(6)
print("Array setelah append:", my_array.tolist())


# --- Array dengan NumPy ---
import numpy as np
np_array = np.array([10, 20, 30, 40, 50])
print("NumPy Array:", np_array)
print("Tambah 100:", np_array + 100)


# --- Matrix operasi dasar ---
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [10, 11, 12]])

print("A + B:\n", A + B)
print("A x B^T:\n", np.dot(A, B.T))


# --- Statistik penjualan + visualisasi ---
import matplotlib.pyplot as plt

penjualan = np.array([120, 155, 130, 160, 185, 145, 170])
print("Rata-rata:", np.mean(penjualan))
print("Median:", np.median(penjualan))
print("Maksimum:", np.max(penjualan))
print("Standar deviasi:", np.std(penjualan))

plt.plot(penjualan, marker="o")
plt.title("Data Penjualan Harian")
plt.xlabel("Hari")
plt.ylabel("Ribuan Rupiah")
plt.show()
```
